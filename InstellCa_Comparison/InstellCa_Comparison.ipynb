# InstellCa comparison: Sadh & Gavassino (2026) irradiance vs this work

**What is compared.** The surface irradiance of a tidally locked planet as a function of the angle from the substellar point, computed by
- **InstellCa_3.0** (Sadh & Gavassino 2026, ApJ 997, 167; code github.com/Mradumay137/InstellCa, MIT license): an unmodified clone lives in `../../Other Repos/InstellCa`. It is run through `InstellCa_3.0_exported.py` in this folder, a copy whose only changes are marked `# MOD` (see its header): the limb-darkening coefficient `u`, the quadrature tolerances and the number of latitude points can be set through environment variables (defaults = original behaviour), and the computed arrays are written to `outfiles/` instead of only being plotted.
- **this work**: `insolation(gamma1, planet, "hyperillum", u_bol)` from `hyperillumination_utils` (closed form in the fully illuminated zone, Gauss–Legendre quadrature in the penumbra), and the point-source `"naive"` model.
- `instellca_transcription.py` (this folder): an independent re-implementation of InstellCa's integral used in Section 5 to locate its bias. It is checked against the actual script output before it is used.

**How the geometry is matched.** InstellCa takes physical inputs (planet radius in Jupiter radii, stellar radius in solar radii, semi-major axis in AU) and converts them with its own constants. Section 1 inverts that conversion so that InstellCa sees exactly the `a/R*` and `r1/r2` of `planet`. The orbit is circular in both codes. Both codes are put on the same stellar luminosity in Section 2, so the comparison isolates the irradiance integral.

**What to look for.** Inside the fully illuminated zone the extended-source irradiance is exactly $L\cos\psi/(4\pi\rho^2)$ ($\rho$: distance from the stellar centre to the surface point), for any limb darkening; InstellCa's own inverse-square curve is this very expression, so it should coincide with `hyperillum` there, while the `naive` model (planet-centre distance) sits lower by the finite-planet-radius factor $(R/(R-r_1))^2$ at the substellar point. Any offset of InstellCa's geometric model from `hyperillum` inside the fully illuminated zone is therefore integration error of InstellCa; it is expected to be flat across the dayside and to grow through the penumbra. Section 3 shows that the numerical floor of this work is many orders of magnitude below that offset, and Section 5 shows that the offset is not a resolution error of InstellCa (its own quadrature and latitude-grid dials do not move it) but a domain error that disappears exactly when the elements beyond the stellar limb / below the horizon are excluded from its integral. The code cells print the numbers; nothing in the text is hard-coded.

```python
import sys, os, subprocess, time
sys.path.insert(0, "../Utilities")
import numpy as np
import matplotlib.pyplot as plt
from hyperillumination_utils import *
from HItargets_utils import *
from paper_plot_utils import paper_figure, paper_theme
import instellca_transcription as it
```

```python
# Target planet and settings (as in the other notebooks)
planet = hi_planets_by_name("TOI 2431 b")   # "K2-141 b" (CUSTOM entry: Malavolta et al. 2018, Barragán et al. 2018), "TOI-2431 b"
instrument = "MIRI/LRS"                       # only used for planet.summary(); the irradiance comparison is bolometric
u_bol = 0.6                                     # not the baseline DEFAULT_U_BOL: InstellCa supports only u = 0.6 and u = 0 (U_RUNS), and the comparison uses its u = 0.6 branch
A_B = DEFAULT_A_B                             # only enters the optional temperature view (Section 6)
U_RUNS = (0.6, 0.0)                           # InstellCa is run for both branches its code supports (u = 0.6 with its Teff x 1.0573 rescaling, and u = 0)
ORDER = 256                                   # Gauss-Legendre order of this work's penumbra quadrature for the comparison (default in insolation(): 64; convergence in Section 3)

planet.summary(print_params=True, eff_wavelength_range=filter_lambda(instrument))
```

## 1. Run InstellCa

`InstellCa_3.0_exported.py` is interactive and must be run with the clone as working directory (it `chdir`s into `./Exoplanet_Catalogs` before anything else). The cell below feeds the prompts of its manual-entry mode (`2`) through `subprocess`, once per value in `U_RUNS`, and the script writes `<planet>_u<u><tag>_instellca.txt` and a `.png` into `outfiles/`. Each run takes a few seconds. Set `RERUN = True` to recompute existing files.

Prompt order of the manual mode: number of planets, catalog choice (`2` = manual; asked because the catalog folder contains several CSV files), planet radius [R$_J$], stellar radius [R$_\odot$], Teff [K], semi-major axis [AU], eccentricity, planet name, image name (overridden by the export). Stellar mass and metallicity are not asked in this mode. The eccentricity must be `0`: for 0 < e < 0.3 InstellCa evaluates at periastron with $a(1-e)$.

```python
INSTELLCA_CLONE = os.path.abspath("../../Other Repos/InstellCa")   # unmodified upstream clone (commit 99225f4); working directory of the run
EXPORT_SCRIPT = os.path.abspath("InstellCa_3.0_exported.py")        # this folder's copy with the array export (see its header)
OUT_DIR = os.path.abspath("outfiles")                               # all run products: InstellCa text files and figures, comparison figures
os.makedirs(OUT_DIR, exist_ok=True)
RERUN = False                                                       # True: run InstellCa again even if the text files exist

# InstellCa's unit constants (from its source): invert them so that a/R* and r1/r2 are exactly the PlanetParams values
RSUN_IC, RJUP_IC, AU_IC = 6.955e8, 6.4e6 * 11.21, 1.496e11
rs1 = 1.0                                        # stellar radius [Rsun]
al1 = planet.R / planet.r2 * RSUN_IC / AU_IC     # semi-major axis [AU]
rp1 = planet.r1 / planet.r2 * RSUN_IC / RJUP_IC  # planet radius [RJ]
prompts = f"1\n2\n{rp1:.10f}\n{rs1}\n{planet.T2_eff:g}\n{al1:.12f}\n0\n{planet.name}\nunused.png\n"

print(f"InstellCa inputs: Rp = {rp1:.10f} RJ, Rs = {rs1} Rsun, Teff = {planet.T2_eff:g} K, a = {al1:.12f} AU, e = 0")
print(f"check: a/R* = {al1 * AU_IC / (rs1 * RSUN_IC):.8f} (PlanetParams {planet.R / planet.r2:.8f}),  r1/r2 = {rp1 * RJUP_IC / (rs1 * RSUN_IC):.8f} (PlanetParams {planet.r1 / planet.r2:.8f})")
```

```python
def instellca_file(u, tag=""):
    return os.path.join(OUT_DIR, planet.name.replace(" ", "_") + f"_u{u:g}{tag}_instellca.txt")

def run_instellca(u, tag="", **knobs):
    """Runs InstellCa_3.0_exported.py for limb-darkening coefficient u. knobs: INSTELLCA_EPSREL / INSTELLCA_EPSABS / INSTELLCA_LIMIT
    (scipy.integrate.quad tolerances; original: scipy defaults) and INSTELLCA_NLAT (latitude points; original: 500)."""
    if os.path.exists(instellca_file(u, tag)) and not RERUN:
        print(f"u = {u}{', ' + tag if tag else ''}: using existing {os.path.basename(instellca_file(u, tag))}")
        return
    env = dict(os.environ, INSTELLCA_U=str(u), INSTELLCA_TAG=tag, MPLBACKEND="Agg", **{k: str(v) for k, v in knobs.items()})
    t0 = time.perf_counter()
    res = subprocess.run([sys.executable, EXPORT_SCRIPT], cwd=INSTELLCA_CLONE, input=prompts, text=True, capture_output=True, env=env)
    tail = res.stdout.strip().splitlines()[-1] if res.stdout.strip() else res.stderr.strip()[-500:]
    print(f"u = {u}{', ' + tag if tag else ''}: exit code {res.returncode} after {time.perf_counter() - t0:.1f} s; InstellCa says: {tail!r}; wrote {os.path.basename(instellca_file(u, tag))}")

for u in U_RUNS:
    run_instellca(u)
```

## 2. Load the InstellCa arrays and put both codes on the same stellar luminosity

The text file holds InstellCa's plotted arrays: the angle from the substellar point in degrees (its grid runs symmetrically from $-s_1$ to $+s_1$ with $s_1$ = floor(90° + arcsin(|R$_s$ − R$_p$|/a)), i.e. slightly short of the true terminator), its geometric (extended-source) model and its inverse-square model, both in kW/m². Two normalisation details are undone here so that InstellCa's star has the luminosity $4\pi r_2^2\sigma T_{2,\mathrm{eff}}^4$ of `planet`: InstellCa uses $\sigma$ = 5.67e-8, and for u = 0.6 it multiplies Teff by 1.0573 to compensate the limb-darkened disk (this conserves the luminosity only to $1.0573^4(1-u/3)$). The inverse-square curve uses the unscaled Teff.

```python
SIGMA_IC = 5.67e-8   # Stefan-Boltzmann constant hard-coded in InstellCa; hyperillumination_utils uses astropy's `sigma`

def lum_factor(u, geometric=True):
    """Luminosity of InstellCa's star in units of 4 pi r2^2 sigma T2_eff^4: the sigma value, and for the geometric model at u = 0.6
    additionally the Teff x 1.0573 rescaling (the inverse-square curve uses the unscaled Teff)."""
    return SIGMA_IC / sigma * ((1.0573 ** 4 * (1 - u / 3)) if (geometric and u == 0.6) else 1.0)

def load_instellca(u, tag=""):
    """Angle from the substellar point [rad, >= 0], InstellCa's geometric (extended-source) irradiance and its inverse-square
    irradiance [W/m^2], both rescaled to the stellar luminosity of `planet`."""
    deg, geo_kw, inv_kw = np.loadtxt(instellca_file(u, tag), unpack=True)
    return np.radians(np.abs(deg)), geo_kw * 1e3 / lum_factor(u), inv_kw * 1e3 / lum_factor(u, geometric=False)

ic = {u: load_instellca(u) for u in U_RUNS}
for u, (gam, E_geo, E_inv) in ic.items():
    i0 = np.argmin(gam)
    print(f"u = {u}: {gam.size} points, |angle| up to {np.degrees(gam.max()):.2f} deg (true terminator {planet.terminator_nightside('deg'):.2f} deg); "
          f"closest to substellar point: {np.degrees(gam[i0]):.2f} deg, geometric {E_geo[i0]/1e3:.4f} kW/m2, inverse-square {E_inv[i0]/1e3:.4f} kW/m2")
```

## 3. This work on the same angles, and its numerical floor

`hyperillum` is closed-form in the fully illuminated zone (no numerics) and a fixed-order Gauss–Legendre quadrature in the penumbra (`insolation_penumbra_hyperillum(..., order)`, default 64). To make sure nothing on this side limits the comparison, the penumbra is evaluated for increasing `order` and compared with an independent reference: `it.exact_E_quad`, adaptive quadrature of the vector-form integral over the exact visible domain (`instellca_transcription.py`, written independently of `hyperillumination_utils`), rescaled to the same luminosity. Two measures are printed per order: the largest relative difference over the penumbra up to 110°, and the largest difference anywhere in the penumbra relative to the substellar irradiance (the relative measure becomes ill-conditioned in the last degree before the terminator, where the irradiance itself goes to zero). The comparison below then uses `ORDER`.

```python
gp1, gp2 = planet.terminator_penumbra(), planet.terminator_nightside()
gam = ic[u_bol][0]
pen = (gam >= gp1) & (gam < gp2)
gam_pen = np.unique(gam[pen])
E_sub = float(insolation(np.array([0.]), planet, "hyperillum", u_bol=u_bol)[0])
geo = it.Geo(planet.R / planet.r2, planet.r1 / planet.r2, planet.T2_eff)
t0 = time.perf_counter()
E_ref = np.array([it.exact_E_quad(float(la), geo, u_bol) for la in gam_pen]) / lum_factor(u_bol)
print(f"independent reference on {gam_pen.size} penumbra angles: {time.perf_counter() - t0:.1f} s")
print(f"{'order':>6} {'max |rel. diff| (<= 110 deg)':>30} {'max |diff| / E_substellar':>28}")
for order in (16, 32, 64, 128, 256, 512, 1024):
    E_o = insolation_penumbra_hyperillum(gam_pen, planet, u_bol=u_bol, order=order)
    inner = gam_pen <= np.radians(110.)
    print(f"{order:6d} {np.max(np.abs(E_o[inner] / E_ref[inner] - 1)):30.1e} {np.max(np.abs(E_o - E_ref)) / E_sub:28.1e}")
print(f"-> comparison below uses order = {ORDER}; the fully illuminated zone is closed-form (checked against the analytic expression in the next cell)")

def hyperillum(gam, u, order=ORDER):
    """insolation(..., 'hyperillum') with a selectable penumbra quadrature order."""
    E = insolation(gam, planet, "hyperillum", u_bol=u)
    m = (gam >= gp1) & (gam < gp2)
    E[m] = insolation_penumbra_hyperillum(gam[m], planet, u_bol=u, order=order)
    return E
```

```python
def analytic_fully_illuminated(gam):
    # L cos(psi) / (4 pi rho^2), rho from the stellar centre to the surface point: exact in the fully illuminated zone for any limb darkening
    rho = np.sqrt(planet.R ** 2 + planet.r1 ** 2 - 2 * planet.R * planet.r1 * np.cos(gam))
    return planet.r2 ** 2 * sigma * planet.T2_eff ** 4 * (planet.R * np.cos(gam) - planet.r1) / rho ** 3

mine = {u: dict(hyperillum=hyperillum(gam, u), naive=insolation(gam, planet, "naive")) for u, (gam, _, _) in ic.items()}

for u, (gam, E_geo, E_inv) in ic.items():
    E_hi = mine[u]["hyperillum"]
    fi, pen = gam < gp1, (gam >= gp1) & (gam < gp2)
    r_fi = E_geo[fi] / E_hi[fi] - 1
    inner = pen & (gam <= np.radians(110.))     # the relative difference diverges in the last degrees before the terminator, where the true irradiance -> 0
    print(f"u = {u}:")
    print(f"   this work vs analytic (fully illuminated zone):            max |rel. diff| = {np.max(np.abs(E_hi[fi] / analytic_fully_illuminated(gam[fi]) - 1)):.1e}")
    print(f"   InstellCa inverse-square vs this work (fully illuminated): max |rel. diff| = {np.max(np.abs(E_inv[fi] / E_hi[fi] - 1)):.1e}")
    print(f"   InstellCa geometric vs this work: fully illuminated zone mean {np.mean(r_fi):+.3%} (min {r_fi.min():+.3%}, max {r_fi.max():+.3%}); "
          f"penumbra up to 110 deg {np.min(E_geo[inner] / E_hi[inner] - 1):+.3%} .. {np.max(E_geo[inner] / E_hi[inner] - 1):+.3%}")
    for d in (float(np.degrees(gp1)), 90., 100., 110.):
        i = np.argmin(np.abs(np.degrees(gam) - d))
        print(f"   at {np.degrees(gam[i]):6.2f} deg: InstellCa geometric / this work - 1 = {E_geo[i] / E_hi[i] - 1:+.3%}")
    i0 = np.argmin(gam)
    print(f"   naive / hyperillum - 1 at the substellar point = {mine[u]['naive'][i0] / E_hi[i0] - 1:+.3%}  (finite-planet-radius factor: {(planet.R / (planet.R - planet.r1)) ** -2 - 1:+.3%})")
```

## 5. Where the InstellCa bias comes from

InstellCa integrates over a **rectangle** in stellar latitude × stellar longitude (its eqs. B7–B9): the longitude limits are the limb longitudes on the stellar equator, the latitude limits the limb latitudes on the central meridian. The visible stellar disk touches that rectangle at four points only; the rectangle's corners are surface elements beyond the limb, which cannot be seen from the planet. Its integrand is the physically correct product (intensity × emission cosine × incidence cosine / distance²), but it takes `abs()` of both cosines, so the invisible elements, whose signed contribution would be negative, are counted positively instead of being dropped. In the penumbra a second domain error adds to this: the lower latitude limit is a horizontal cut through the (curved) horizon (eq. B10), and elements below the horizon inside the rectangle are again rectified by the `abs()`. The paper attributes the < 1 % discrepancy to numerical error "that can be reduced if the latitude arrays are made more densely spaced".

Three tests, all against this work (which Section 3 pins to the analytic value / the independent reference):
1. **InstellCa's own dials.** The exported script is re-run with scipy's quadrature tolerances tightened by orders of magnitude (its default absolute tolerance of 1.5e-8 is never binding for an integrand of order 1e-22 in SI units, so by default scipy stops after its first 21-point estimate) and with 4× denser latitude arrays. If the offset were a resolution error, it would move.
2. **The transcription.** `it.instellca_E` reproduces the script's numbers (checked first). Then the same integrand and quadrature are evaluated with one change at a time: (a) InstellCa's rectangle with its `abs()` (= InstellCa), (b) the rectangle but with the elements outside the true domain set to zero, (c) the true domain (visible cap cut by the horizon). If the offset comes from the domain, (b) and (c) must agree with this work.
3. **The geometry**, drawn: which stellar surface elements InstellCa integrates over versus which ones actually illuminate the point. The invisible area inside the rectangle is a sizeable fraction of the visible disk, but it lies just beyond the limb where the emission cosine is near zero, so its rectified contribution to the irradiance is only of order a percent (and smaller for u = 0.6 than for u = 0, since limb darkening suppresses exactly that region).

```python
# 5.1 InstellCa's own dials: quadrature tolerances and latitude-grid density (same u as the paper figure)
# the tight-tolerance run costs ~1 s per latitude point (adaptive subdivision around the abs() kinks), so it is done on 120 points
DIALS = {"": {}, "_quad1e-8_nlat120": dict(INSTELLCA_EPSREL=1e-8, INSTELLCA_EPSABS=0, INSTELLCA_LIMIT=200, INSTELLCA_NLAT=120), "_nlat2000": dict(INSTELLCA_NLAT=2000)}
for tag, knobs in DIALS.items():
    run_instellca(u_bol, tag, **knobs)
dial = {tag: load_instellca(u_bol, tag) for tag in DIALS}
print(f"\nInstellCa (u = {u_bol:g}) relative to this work:")
print(f"{'run':>14} {'points':>7} {'fully illuminated zone: mean':>30} {'spread (max-min)':>18} {'at 90 deg':>11} {'at 100 deg':>11}")
for tag, (g_t, geo_t, _) in dial.items():
    E_hi_t = hyperillum(g_t, u_bol)
    r = geo_t / E_hi_t - 1
    fi_t = g_t < gp1
    i90, i100 = np.argmin(np.abs(np.degrees(g_t) - 90)), np.argmin(np.abs(np.degrees(g_t) - 100))
    print(f"{tag or 'as published':>14} {g_t.size:7d} {np.mean(r[fi_t]):+30.4%} {np.ptp(r[fi_t]):18.4%} {r[i90]:+11.3%} {r[i100]:+11.3%}")
```

```python
# 5.2 Transcription: (i) does it reproduce the script? (ii) what happens when the integration domain is corrected?
gam, E_geo, E_inv = ic[u_bol]
idx = np.argsort(gam)[::25]                                         # every 25th grid point
E_tr = np.array([it.instellca_E(float(gam[i]), geo, u_bol) for i in idx]) / lum_factor(u_bol)
print(f"transcription vs InstellCa script output at {idx.size} angles: max |rel. diff| = {np.max(np.abs(E_tr / E_geo[idx] - 1)):.1e}  (the inputs are rounded to 10 digits)")

test_deg = [0.5, 30., 55., float(np.degrees(gp1)) + 2, 80., 90., 100., 110.]
print(f"\n{'angle [deg]':>11} | {'InstellCa script':>16} {'(a) rectangle + abs':>20} {'(b) rectangle, masked':>22} {'(c) true domain':>16}   (each: value / this work - 1)")
variants = {}
for d in test_deg:
    la = float(np.radians(d))
    E_hi_d = float(hyperillum(np.array([la]), u_bol)[0])
    i_near = np.argmin(np.abs(gam - la))
    va = it.instellca_variant(la, geo, u_bol, "box", True) / lum_factor(u_bol)
    vb = it.instellca_variant(la, geo, u_bol, "box_masked") / lum_factor(u_bol)
    vc = it.instellca_variant(la, geo, u_bol, "exact") / lum_factor(u_bol)
    variants[d] = (va / E_hi_d - 1, vb / E_hi_d - 1, vc / E_hi_d - 1)
    script = E_geo[i_near] / mine[u_bol]["hyperillum"][i_near] - 1
    print(f"{d:11.2f} | {script:+16.4%} {va / E_hi_d - 1:+20.4%} {vb / E_hi_d - 1:+22.2e} {vc / E_hi_d - 1:+16.1e}")
print("(the script column is the nearest grid point, computed with scipy's default tolerances; (a)-(c) use epsrel = 1e-10)")
```

```python
PAPER_MODE = False
# 5.3 Convergence figure: InstellCa with its dials vs. InstellCa with the integration domain corrected, all relative to this work
fig, ax = plt.subplots(figsize=(7.5, 4.8))
# the three InstellCa runs lie on top of each other: the 2000-point run is drawn as a line, the published 500-point run as dots, the tight-tolerance run as crosses on top
g_t, geo_t, _ = dial["_nlat2000"]
ax.plot(np.degrees(g_t), 100 * (geo_t / hyperillum(g_t, u_bol) - 1), "-", lw=1.2, color="C4", label="InstellCa, 2000 latitude points (the paper's proposed remedy)")
g_t, geo_t, _ = dial[""]
ax.plot(np.degrees(g_t), 100 * (geo_t / hyperillum(g_t, u_bol) - 1), ".", ms=3, color="C3", label="InstellCa as published (scipy default tolerances, 500 latitude points)")
g_t, geo_t, _ = dial["_quad1e-8_nlat120"]
ax.plot(np.degrees(g_t), 100 * (geo_t / hyperillum(g_t, u_bol) - 1), "x", ms=6, mew=1.2, color="k", label="InstellCa, quadrature tolerance 1e-8 (absolute 0), limit 200, 120 latitude points")
dd = np.array(test_deg)
ax.plot(dd, 100 * np.array([variants[d][1] for d in test_deg]), "s", ms=9, mfc="none", mew=1.5, color="C0", label="InstellCa integrand on its rectangle, invisible elements removed")
ax.plot(dd, 100 * np.array([variants[d][2] for d in test_deg]), "o", ms=5, color="C0", label="InstellCa integrand on the true visible domain")
ax.axhline(0, color="gray", lw=0.8)
shade_zones(ax, label_y=2.45, fontsize=8)
ax.set_xlim(0, np.degrees(gp2) + 4); ax.set_ylim(-0.3, 2.5)
ax.set_xlabel("angle from the substellar point [deg]"); ax.set_ylabel(f"InstellCa variant / this work − 1 [%]\n(both extended source, u = {u_bol:g})")
ax.set_title(f"{planet.name}: the InstellCa offset is a domain error, not a resolution error")
ax.legend(fontsize=7.5, loc="upper left", bbox_to_anchor=(0.0, 0.93))
fig.tight_layout()
fig.savefig(os.path.join(OUT_DIR, f"InstellCa_convergence_{planet.name.replace(' ', '_')}.pdf"))
fig.savefig(os.path.join(OUT_DIR, f"InstellCa_convergence_{planet.name.replace(' ', '_')}.png"), dpi=200)
paper_figure(f"instellca_convergence_{planet.name.replace(' ', '')}", PAPER_MODE, fig=fig)
plt.show()
```

```python
PAPER_MODE = False
# 5.4 The integration domains on the stellar sphere, for one point in the fully illuminated zone and one in the penumbra
from matplotlib.colors import ListedColormap
from matplotlib.patches import Patch
fig, axes = plt.subplots(1, 2, figsize=(12, 5.4))
for ax, d in zip(axes, (30., 95.)):
    m = it.domain_masks(float(np.radians(d)), geo, n=700)
    cat = np.zeros(m["true"].shape, dtype=int)                   # 0: neither
    cat[m["true"] & m["box"]] = 1                                 # integrated by InstellCa and truly illuminating
    cat[m["box"] & ~m["true"]] = 2                                # integrated by InstellCa but invisible (beyond the limb or below the horizon) -> counted via abs()
    cat[m["true"] & ~m["box"]] = 3                                # illuminating but outside InstellCa's rectangle -> missed
    cmap = ListedColormap(["white", "#9ecae1", "#e34a33", "#fdae6b"])
    ax.pcolormesh(np.degrees(m["x"]), np.degrees(m["th"]), cat, cmap=cmap, vmin=-0.5, vmax=3.5, shading="auto", rasterized=True)
    ax.contour(np.degrees(m["x"]), np.degrees(m["th"]), m["cap"].astype(float), levels=[0.5], colors="k", linewidths=0.8)
    ax.contour(np.degrees(m["x"]), np.degrees(m["th"]), m["above"].astype(float), levels=[0.5], colors="k", linewidths=0.8, linestyles="--")
    ax.add_patch(plt.Rectangle((-np.degrees(m["y3"]), np.degrees(m["ll"])), 2 * np.degrees(m["y3"]), np.degrees(m["ul"] - m["ll"]), fill=False, ec="C3", lw=1.5))
    frac_spur = (cat == 2).sum() / max((cat == 1).sum(), 1); frac_miss = (cat == 3).sum() / max((cat == 1).sum(), 1)
    ax.set_title(f"surface point {d:g}° from the substellar point ({'fully illuminated zone' if np.radians(d) < gp1 else 'penumbra'})\n"
                 f"red = {frac_spur:.0%} of the blue area, but only ~1 % of its flux (grazing emission); orange = {frac_miss:.1%}", fontsize=8.5)
    ax.text(0, np.degrees(m["ul"]) - 1.5, "beyond the limb:\ninvisible, yet integrated", ha="center", va="top", fontsize=7.5, color="darkred")
    ax.text(0, 0.5 * np.degrees(m["ll"] + m["ul"]), "visible disk", ha="center", va="center", fontsize=9, color="navy")
    if np.radians(d) >= gp1:
        ax.text(0, np.degrees(m["ll"]) - 0.6, "below the horizon (horizontal cut, eq. B10)", ha="center", va="top", fontsize=7.5, color="darkred")
    ax.set_xlabel("stellar longitude from the sub-planet meridian [deg]"); ax.set_ylabel("stellar latitude [deg]"); ax.set_aspect("equal")
fig.legend(handles=[Patch(fc="#9ecae1", label="correct: element illuminates the point and is integrated by InstellCa"),
                        Patch(fc="#e34a33", label="the error: element does not illuminate the point (beyond its limb / below its horizon),\nbut InstellCa integrates it, with positive sign because of the abs()"),
                        Patch(fc="#fdae6b", label="element illuminates the point but is not integrated (none here)"),
                        plt.Line2D([], [], color="C3", lw=1.5, label="InstellCa integration rectangle (eqs. B7–B10)"),
                        plt.Line2D([], [], color="k", lw=0.8, label="stellar limb as seen from the point"),
                        plt.Line2D([], [], color="k", lw=0.8, ls="--", label="horizon of the point")], fontsize=7.5, loc="lower center", ncol=2, frameon=False)
fig.suptitle(f"{planet.name}: InstellCa's integration domain vs the stellar surface that actually illuminates the point", fontsize=10)
fig.tight_layout(rect=(0, 0.17, 1, 1))
fig.savefig(os.path.join(OUT_DIR, f"InstellCa_domain_{planet.name.replace(' ', '_')}.pdf"))
fig.savefig(os.path.join(OUT_DIR, f"InstellCa_domain_{planet.name.replace(' ', '_')}.png"), dpi=200)
paper_figure(f"instellca_domain_{planet.name.replace(' ', '')}", PAPER_MODE, fig=fig, tight=False)
plt.show()
```

## 7. Paper figure (Appendix A.4)

Two panels for the appendix, in the paper's notation ($\gamma$, $I$, $u_{\rm bol}$). (a): the integration domain of InstellCa for a surface point in the fully illuminated zone. (b): the published treatments that include limb darkening, relative to this work, for both limb-darkening values (colour = $u_{\rm bol}$):
- InstellCa (points, on its own grid, all above zero) and its inverse-square comparison curve (dashed, drawn only where it is positive);
- the closed-form approximation of Kopal (1954) (lines, all below zero): his eqs. 43–48 with the sign of $\lambda$ corrected (`insolation_penumbra_hyperillum_approx`) in the penumbra, his exact eq. 34 in the fully illuminated zone.

The two treatments are labelled on the plot, and the legend keys $u_{\rm bol}$ and the dashed curve. The y axis is symmetric-logarithmic (linear within ±0.1 %), because the InstellCa dayside offset and Kopal's error near the terminator differ by about two orders of magnitude. Kang et al. (2023) and Carter et al. (2024) are not drawn: for $u_{\rm bol} = 0$ both converge to this work and would lie on the zero line. The panel titles are for the notebook only; `paper_figure(..., titles="remove")` drops them (the caption carries that information) and the panels are labelled (a) and (b). Saved as `outfiles/InstellCa_paper_<planet>.pdf` and, with `PAPER_MODE = True`, to the paper storage as `InstellCa_Comparison.pdf`.
