```python
import batman
import numpy as np
import matplotlib.pyplot as plt
import utils
```

```python
period=1.                               #orbital period
rp = 0.1                                    #planet radius (in units of stellar radii)
a = 5.                                     #semi-major axis (in units of stellar radii)
incl = 89.                                  #orbital inclination (in degrees)
ecc = 0.95                                    #eccentricity
w = 0.                                      #longitude of periastron (in degrees)
fp = 0.01                                   #planet-to-star flux ratio (eclipse depth)

t = np.linspace(-period/4, period*3/4, 10000)
```

```python
w1=0.
plt.figure(figsize=(16, 6))
colors = ['b', 'g', 'r', 'c', 'k']
for i, ecc1 in enumerate([0.7, 0.8, 0.9, 0.95, 0.98]):
    flux=utils.flux_eclipse(period, rp, a, incl, ecc1, w1, fp, t)
    plt.plot(t, flux, label=ecc1, color=colors[i])
    #plt.vlines(utils.eclipse_time_offset(period, ecc1, w1), ymin=flux.min(), ymax=flux.max(), color=colors[i], linestyles='dashed')
plt.xlabel("Time from central transit")
plt.xlim([period*0.4, period*0.6])
plt.ylabel("Norm. flux")
plt.legend()
plt.show()
```
