```python
import batman
import numpy as np
import matplotlib.pyplot as plt
import phasecurve_utils
```

```python
period=1.                                   #orbital period
rp = 0.1                                    #planet radius (in units of stellar radii)
a = 15.                                     #semi-major axis (in units of stellar radii)
incl = 89.                                  #orbital inclination (in degrees)
ecc = 0.                                    #eccentricity
w = 90.                                     #longitude of periastron (in degrees)
fp = 0.001                                  #planet-to-star flux ratio (eclipse depth)
rs = 1                                      # stellar radius (in units of solar radii)

offset = 10                                 # hotspot offset delta (atmospheric winds redistributing heat east of substellar point, in degrees)

t = np.linspace(-period/4, period*3/4, 10000)
```

### Transit

```python
plt.figure(figsize=(16, 6))
plt.plot(t, phasecurve_utils.flux_transit(period, rp, a, incl, ecc, w, t))
plt.xlabel("Time from central transit")
plt.ylabel("Norm. flux")
plt.show()
```

### Eclipse

```python
plt.figure(figsize=(16, 6))
plt.plot(t, phasecurve_utils.flux_eclipse(period, rp, a, incl, ecc, w, fp, t))
plt.xlabel("Time from central transit")
plt.ylabel("Norm. flux")
plt.show()
```

### Eclipse + Planetary Phases

```python
phase=(np.cos(2*np.pi * (t/period - 0.5) + np.radians(offset))+1)/2
plt.figure(figsize=(16, 6))
plt.plot(t, (phasecurve_utils.flux_eclipse(period, rp, a, incl, ecc, w, fp, t)-1)*phase)
plt.xlabel("Time from central transit")
plt.ylabel("Norm. flux")
plt.show()
```

### Transit + Eclipse + Planetary Phases

```python
plt.figure(figsize=(16, 6))
plt.plot(t, phasecurve_utils.flux_transit(period, rp, a, incl, ecc, w, t)+(phasecurve_utils.flux_eclipse(period, rp, a, incl, ecc, w, fp, t)-1)*phase)
plt.xlabel("Time from central transit")
plt.ylabel("Norm. flux")
plt.show()
```
