```python
import numpy as np
import matplotlib.pyplot as plt
import phasecurve_utils
```

```python
# Data from K2-141b (Malavolta et al. 2018, Barragán et al. 2018)
period=0.2803                               #orbital period (days)
rp = 0.02                                   #planet radius (in units of stellar radii)
a = 0.007*1.5e11/(0.68*7e8)                 #semi-major axis (in units of stellar radii)
incl = 86.3                                 #orbital inclination (in degrees)
ecc = 0.                                    #eccentricity
w = 90.                                     #longitude of periastron (in degrees)
fp = 142e-6                                 #planet-to-star flux ratio / eclipse depth (integrated day-side flux)
rs = 0.68                                   # stellar radius (in units of solar radii)

A=1.                                        # Phase Curve Amplitude ( A = 1 - F_p,min / F_p,max = (T_night / T_day)^4 )
```

```python
t, flux = phasecurve_utils.PhaseCurve(period, rp, a, incl, ecc, w, fp, rs, t=np.linspace(0, period, 10000))
```

```python
plt.figure(figsize=(16, 6))
plt.plot(t/period, flux*1e6)
plt.axhline(1e6, color='r', linestyle='--', linewidth=0.5)
plt.xlabel("Phase")
plt.ylabel("Norm. flux (ppm)")
plt.grid()
plt.show()
```
