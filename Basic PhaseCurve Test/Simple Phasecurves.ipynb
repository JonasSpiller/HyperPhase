```python
import numpy as np
import matplotlib.pyplot as plt
import utils
```

```python
# Data from K2-141b (Malavolta et al. 2018, Barragán et al. 2018)
period=0.2803                               #orbital period (days)
rp = 0.02                                   #planet radius (in units of stellar radii)
a = 0.007*1.5e11/(0.68*7e8)                 #semi-major axis (in units of stellar radii)
incl = 86.3                                 #orbital inclination (in degrees)
ecc = 0.                                    #eccentricity
w = 90.                                     #longitude of periastron (in degrees)
fp = 142e-6                                 #planet-to-star flux ratio / eclipse depth

```

```python
# Value to be fitted
offset = 0                                  # hotspot offset delta (atmospheric winds redistributing heat east of substellar point)
```

```python
t, flux = utils.PhaseCurve(period, rp, a, incl, ecc, w, fp, offset, np.linspace(0, period, 10000))
```

```python
plt.figure(figsize=(16, 6))
plt.plot(t/period, flux*1e6)
plt.xlabel("Phase")
plt.ylabel("Norm. flux (ppm)")
plt.grid()
plt.show()
```
